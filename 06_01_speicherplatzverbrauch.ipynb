# Untersuchung Speicherplatzverbrauch Arrays vs. der Python-Datentyp Liste

Zunächst erstellen wir eine leere Liste und schauen wie groß sie ist

```python
import sys

meine_liste = []
print(f"Leer: {sys.getsizeof(meine_liste)} Bytes")
```

Lassen Sie uns einige Elemente hinzufügen und dabei prüfen wie die Größe der Liste anwächst. Führen Sie die nächste Zelle ruhig häufiger nacheinander aus.


Achten Sie darauf wie die Speichergröße steigt.

```python
for i in range(10):
    meine_liste.append(i)
    print(f"Elemente: {len(meine_liste)}, Speicher: {sys.getsizeof(meine_liste)} Bytes")
```

Python reserviert bei der Listenerstellung direkt Speicher für weitere Elemente. Dadurch muss nicht bei jedem neu angehängten Element erneut Speicher angefragt werden --> Prinzip der Overallocation

Um einen fairen Vergleich zwischen Arrays und Listen durchzuführen, erstellen wir eine Liste mit 10 "Nullen" und ein Array mit 10 "Nullen" und vergleichen diese

```python
# ein array als Liste
meine_liste = [0] * 10
print(meine_liste)
print(f"Array als Liste: {sys.getsizeof(meine_liste)} Bytes")
```

Um ein Array in Python zu erzeugen, benötigen wir das Package "numpy"

```python
import numpy as np
arr = np.zeros(10, dtype=int)  # oder np.array([...])
print(arr)
print(f"\nSpeicher numpyarray: {sys.getsizeof(arr)} Bytes")
```

Wenn wir nun einen integer-Wert ändern, sollte die Größe gleich sein. Lassen Sie uns prüfen, ob das stimmt

```python
arr[2] = 42
print(f"Das Array: {arr}")
print(f"\nSpeicher numpyarray: {sys.getsizeof(arr)} Bytes")
```

Hatte Ich Ihnen nicht erzählt, dass Arrays speichereffizienter sind? Warum ist dann eigentlich die Liste kleiner als das array?

Wir müssen richtig messen...

```python
# Für NumPy: Daten + Overhead
print(f"NumPy Daten: {arr.nbytes} Bytes")
print(f"NumPy Overhead: {sys.getsizeof(arr) - arr.nbytes} Bytes")
```

```python
# Für Listen: Container + alle Integer-Objekte
list_total = sys.getsizeof(meine_liste) + sum(sys.getsizeof(item) for item in meine_liste)
print(f"Liste Total (mit Objekten): {list_total} Bytes")
```

WICHTIG: Für dynamisches Wachstum sind Listen besser! Arrays müssen immer neu erstellt werden, wenn ein bei einem bereits "vollen" Array ein weiteres Element hinzugefügt werden soll.


NumPy-Arrays sind für FESTE Größen optimiert!
Für dynamisches Wachstum: Python-Liste verwenden, dann zu Array konvertieren, wenn weitere Performance kritisch ist
