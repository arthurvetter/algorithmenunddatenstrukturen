## 1. Motivation: eingebaute, array-basierte Listen-Laufzeiten

Was sind die Laufzeitkomplexitäten einiger gängiger Operationen zum Aktualisieren einer array-basierten Liste?

```python
import matplotlib.pyplot as plt
import numpy as np
from timeit import timeit
%matplotlib inline
```

### Indexbasierte Aktualisierung

```python
size = 10_000

ts = [timeit('lst[{}] = None'.format(n),
             setup=f'lst = list(range({size}))',
             number=100)
      for n in range(0, size, 10)]

plt.plot(ts, 'ro');
```

### Anhängen

```python
ns = np.linspace(100, 10000, 50, dtype=int)

ts = [timeit('lst.append(None)',
             setup=f'lst = list(range({n}))',
             number=1000)
      for n in ns]

plt.plot(ns, ts, 'ro');
```

### Einfügen & Löschen

```python
ns = np.linspace(100, 10000, 50, dtype=int)

# timing insertion at front
ins_ts = [timeit('lst.insert(0, None)',
                 setup=f'lst = list(range({n}))',
                 number=100)
          for n in ns]

# timing deletion at front
del_ts = [timeit('del lst[0]',
                 setup=f'lst = list(range({n}))',
                 number=100)
          for n in ns]

plt.plot(ns, ins_ts, 'go')
plt.plot(ns, del_ts, 'ro');
```

### Verkettung

```python
def concat1(lst1, lst2):
    for x in lst2:
        lst1.append(x)
    return lst1

def concat2(lst1, lst2):
    lst1.extend(lst2)
    return lst1

def concat3(lst1, lst2):
    return lst1 + lst2
```

```python
ns = np.linspace(100, 10000, 50, dtype=int)

ts = [[timeit('{}(lst1, lst2)'.format(fn),
              setup=f'lst1 = list(range({n})); lst2 = list(range({n}))',
              number=100,
              globals=globals())
       for n in ns] 
      for fn in ('concat1', 'concat2', 'concat3')]

plt.plot(ns, ts[0], 'ro')
plt.plot(ns, ts[1], 'go')
plt.plot(ns, ts[2], 'bo');
```

### Fazit

Während indexbasierte Updates und Anhänge $O(1)$ sind, sind Operationen, die möglicherweise die relativen Positionen von Elementen en masse ändern müssen (z. B. Einfügen, Löschen, Verkettung), $O(N)$.

Dies liegt an der **monolithischen, zusammenhängenden Speicherzuweisung** des Arrays, das verwendet wird, um Referenzen auf Elemente in der Liste zu halten.

## 2. Wunschliste

Für Anwendungen, die häufig Operationen wie Einfügungen und Löschungen durchführen (die in einer array-basierten Liste ineffizient sind), möchten wir einen verbesserten Speichermechanismus, der:

- **keine** monolithische, zusammenhängende Speicherzuweisung erfordert,
- eine flexible und effiziente Umstrukturierung von Teilstrukturen unterstützt (z. B. für Verkettung und Verschiebung),
- und es uns ermöglicht, darauf aufbauend weitere häufig benötigte Operationen (z. B. für Indexierung, Iteration, Suche) zu implementieren.

### 4.2 Binärbaum

```python
class BinaryLink:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
# an "expression tree" representing the arithmetic expression ((5+3)*(8-4))
t = BinaryLink('*')
t.left = BinaryLink('+')
t.left.left  = BinaryLink('5')
t.left.right = BinaryLink('3')
t.right = BinaryLink('-')
t.right.left  = BinaryLink('8')
t.right.right = BinaryLink('4')
```

```python
def print_expr_tree(t):
    if t:
        if not t.val.isdigit():
            print('(', end='')
        print_expr_tree(t.left)
        print(t.val, end='')
        print_expr_tree(t.right)
        if not t.val.isdigit():
            print(')', end='')
```

```python
print_expr_tree(t)
```

### 4.3 N-ärer Baum

```python
class NaryLink:
    def __init__(self, val, n=2):
        self.val = val
        self.children = [None] * n
        
    def __getitem__(self, idx):
        return self.children[idx]
    
    def __setitem__(self, idx, val):
        self.children[idx] = val
        
    def __iter__(self):
        for c in self.children:
            yield c
```

```python
root = NaryLink('Kingdoms', n=5)

root[0] = NaryLink('Animalia', n=35)
root[1] = NaryLink('Plantae', n=12)
root[2] = NaryLink('Fungi', n=7)
root[3] = NaryLink('Protista', n=5)
root[4] = NaryLink('Monera', n=5)

root[2][0] = NaryLink('Chytridiomycota')
root[2][1] = NaryLink('Blastocladiomycota')
root[2][2] = NaryLink('Glomeromycota')
root[2][3] = NaryLink('Ascomycota')
root[2][4] = NaryLink('Basidiomycota')
root[2][5] = NaryLink('Microsporidia')
root[2][6] = NaryLink('Neocallimastigomycota')

def tree_iter(root):
    if root:
        yield root.val
        for c in root:
            yield from tree_iter(c)
```

```python
for x in tree_iter(root):
    print(x)
```
