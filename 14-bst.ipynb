## Binärbaum: Definition

- Ein *Binärbaum* ist eine Struktur, die entweder leer ist oder aus einem *Wurzel*-Knoten besteht, der einen Wert und Verweise auf einen linken und rechten *Teilbaum* enthält, welche selbst Binärbäume sind.

Benennung der Knoten:
- Der einzelne Knoten in einem Binärbaum ohne Elternknoten ist der Wurzelknoten des Baumes
- Wir sagen, dass ein gegebener Knoten der *Elternteil* seiner linken und rechten *Kind*-Knoten ist; Knoten mit demselben Elternteil werden *Geschwister* genannt
- Hat ein Knoten keine Kinder, nennen wir ihn einen *Blatt*-Knoten; andernfalls nennen wir ihn einen *inneren* Knoten

Metriken des Binärbaums (Hinweis: alternative Definitionen werden manchmal verwendet!):

- Die *Tiefe* eines Knotens ist die Anzahl der Knoten vom Wurzelknoten des Baumes bis zu diesem Knoten (einschließlich)
- Die *Höhe* eines Knotens ist die Anzahl der Knoten auf dem längsten Pfad von diesem Knoten bis zu einem Blatt (einschließlich)

Kategorisierung von Binärbäumen:

- In einem *vollständigen* Binärbaum hat jeder Knoten entweder 0 oder 2 Kinder
- In einem *kompletten* Binärbaum sind alle Ebenen bis auf die letzte gefüllt, und die letzte Ebene wird von links nach rechts aufgefüllt
- In einem *perfekten* Binärbaum haben alle Blätter die gleiche Tiefe
- In einem *balancierten* Binärbaum ... ?

## Binärer Suchbaum (BSTree): Definition

- Ein *binärer Suchbaum* ist ein binärer Baum, bei dem der Wert in jedem Knoten:
    - *größer als* alle Werte im linken Teilbaum ist, und
    - *kleiner als* alle Werte im rechten Teilbaum ist
