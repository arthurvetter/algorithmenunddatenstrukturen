## 1. Rekursion

Rekursive Funktionen rufen sich direkt oder indirekt selbst auf.

Rekursive Lösungen sind anwendbar, wenn ein Problem in leichter zu lösende Teilprobleme zerlegt werden kann, die dem ursprünglichen Problem ähneln und deren Lösungen anschließend kombiniert werden können.

z. B. die Berechnung des Gesamtpreises einer Reihe von verschachtelten Einkaufstaschen mit Artikeln:

````python
class Bag:
    def __init__(self, price, *contents):
        self.price = price
        self.contents = contents

    def __repr__(self) -> str:
        return '(' + str(self.price) + ' ' + ', '.join(repr(x) for x in self.contents) + ')'
````

````python
bag1 = Bag(10)
````

````python
bag2 = Bag(5, 
           Bag(3), 
           Bag(7))
print(bag2)
````

````python
bag3 = Bag(5, 
           Bag(4, 
               Bag(3)), 
           Bag(2))
````

````python
bag4 = Bag(0, 
           Bag(5), 
           Bag(10), 
           Bag(3, 
               Bag(2), 
               Bag(100)), 
           Bag(9, 
               Bag(2, 
                   Bag(25))))
````

````python
def price(bag):
    print("Price-Function Called")
    total = bag.price
    for b in bag.contents:
        total += price(b)
    return total
````

````python
price(bag4)
````

### Der Call-Stack
Wenn eine Funktion aufgerufen wird, wird im sogenannten Call Stack ein entsprechender Stack-Frame angelegt. Dieser Stack-Frame speichert unter anderem:

- die Argumente der Funktion
- weitere lokale Variablen (falls vorhanden)
- die Adresse im aufrufenden Code, an der die Ausführung fortgesetzt werden soll

Wenn die Funktion zurückkehrt, wird ihr Stack-Frame wieder entfernt. Der Call Stack ist eine LIFO-Datenstruktur („Last-In, First-Out“).

Das gilt sowohl für nicht-rekursive als auch für rekursive Funktionen. Bei rekursiven Funktionen kann sich eine lange Kette verschachtelter Aufrufe aufbauen, bevor mit dem Zurückkehren begonnen wird. Gibt es keine Abbruchbedingung, wächst diese Kette unbegrenzt und führt schließlich zu einem Stack Overflow.

Durch das Setzen eines Breakpoints im Debugger können wir den Call Stack in Aktion beobachten.

````python
import sys
sys.setrecursionlimit(500)
````

````python
def silly_rec(n):
    print(n)
    if n == 0:
        return 0
    else:
        silly_rec(n - 1)
````

````python
silly_rec(10)
````

## 2. Rekursive „Formen“

### Lineare Rekursion

#### Beispiel: Summe (für Eingabe ≥ 0)

$sum(n) = \begin{cases}
        0 & \text{wenn}\ n=0 \\
        n + sum(n-1) & \text{wenn}\ n>0
      \end{cases}$


Zum Beispiel, 

$
\begin{align}
sum(5) & = 5 + sum(4)\\
       & = 5 + 4 + sum(3)\\
       & = 5 + 4 + 3 + sum(2)\\
       & = 5 + 4 + 3 + 2 + sum(1)\\
       & = 5 + 4 + 3 + 2 + 1 + sum(0)\\
       & = 5 + 4 + 3 + 2 + 1 + 0
\end{align}
$

````python
def sum(n):
    if n == 0:
        return 0
    else:
        return n + sum(n-1)
````

````python
sum(5)
````

#### Beispiel: Fakultät

$n! = \begin{cases}
        1 & \text{wenn}\ n=0 \\
        n \cdot (n-1)! & \text{wenn}\ n>0
      \end{cases}$

````python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
````

````python
factorial(5)
````

#### Beispiel: Binäre Suche mit Rekursion

Die binäre Suche eignet sich besonders gut für rekursive Implementierungen, da das Problem bei jedem Schritt in zwei kleinere Teilprobleme aufgeteilt wird. Bei jeder Rekursion wird der Suchbereich halbiert, bis das gesuchte Element gefunden ist oder der Bereich leer ist. 

````python
def bin_search(x, lst):
    print(lst)
    if len(lst) == 0:
        return False
    else:
        mid = len(lst) // 2
        if lst[mid] == x:
            return True
        elif lst[mid] < x:
            return bin_search(x, lst[mid+1:])
        else:
            return bin_search(x, lst[:mid])
````

````python
bin_search(25, list(range(100)))
````

````python
bin_search(99, list(range(100)))
````

````python
bin_search(100, list(range(100)))
````

### Baumrekursion

#### Beispiel: Fibonacci-Zahlen

$fib(n) = \begin{cases}
            0 & \text{falls}\ n=0 \\
            1 & \text{falls}\ n=1 \\
            fib(n-1) + fib(n-2) & \text{ansonsten}
          \end{cases}$
          
d.h., 0, 1, 1, 2, 3, 5, 8, 13, 21, ...

````python
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n-1) + fib(n-2)
````

````python
fib(3)
````

````python
[fib(i) for i in range(15)]
````

Natürlich kann die Fibonacci-Folge auch iterativ berechnet werden. Ab der dritten Zahl ist die aktuelle Zahl einfach die Summe der beiden vorherigen Zahlen.

````python
def fib_iter(n):
    f0 = 0
    f1 = 1
    for _ in range(n):
        f0, f1 = f1, f0 + f1
    return f0
````

````python
fib_iter(20)
````

````python
fib_iter(200)
````

````python
import timeit
import matplotlib.pyplot as plt

ns = range(20)
tRec = [timeit.timeit(stmt=f'fib({n})', 
                    globals=globals(),
                    number=100)
      for n in ns]
tIt = [timeit.timeit(stmt=f'fib_iter({n})', 
                    globals=globals(),
                    number=100)
      for n in ns]

plt.plot(ns, tRec, 'or');
plt.plot(ns, tIt, 'ob')
````

# Warum ist rekursiv so viel langsamer?  
Hinweis: Gehe den Aufrufstapel Aufruf für Aufruf des rekursiven Algorithmus an einem kleinen Beispiel durch:  
``` 
                        fib(4)
            fib(3)                 fib(2)
      fib(2)      fib(1)       fib(1)  fib(0)
fib(1)    fib(0)
```

Der iterative Ansatz hat also $O(N)$, während unser rekursiver Ansatz ein exponentielles Wachstum aufweist (genauer gesagt $O(\phi^n)$, wobei $\phi = \frac{1+\sqrt{5}}{2}$ ist).

#### Beispiel: Turm von Hanoi

Aufbau: drei Stäbe, mit einer oder mehreren Scheiben unterschiedlicher Größe, die alle auf einem Stab gestapelt sind, kleinste (oben) bis größte (unten). Zum Beispiel:

         ||          ||          ||     
         ==          ||          ||     
        ====         ||          ||     
       ======        ||          ||     
    ------------------------------------
    
Ziel: Alle Scheiben nacheinander auf einen anderen Stab zu bewegen, wobei die Regeln sind, dass (1) nur kleinere Scheiben auf größere gestapelt werden dürfen und (2) nur die oberste Scheibe eines Stapels auf einen anderen Stab bewegt werden kann.

Für drei Scheiben, wie oben gezeigt, würden wir die folgende Reihenfolge ausführen, um den Stapel auf den rechten Stab zu bewegen. Die Stäbe werden mit L (links), M (Mitte), R (rechts) abgekürzt:
1. Bewege die kleine Scheibe (0) von L nach R
2. Bewege die mittlere Scheibe (1) von L nach M
3. Bewege 0 von R nach M (R ist jetzt leer)
4. Bewege die große Scheibe (2) von L nach R
5. Bewege 0 von M nach L
6. Bewege 1 von M nach R
7. Bewege 0 von L nach R (fertig)

Kannst du die Reihenfolge finden, die nötig ist, um einen Stapel von 4 Scheiben von einem Stab auf einen anderen zu bewegen? 5 Scheiben? Eine beliebige Anzahl von Scheiben?

````python
height = 3
towers = [[] for _ in range(3)]
towers[0] = list(range(height, 0, -1))
````

````python
towers
````

````python
from time import sleep
from IPython.display import clear_output

def display():
    clear_output(True)
    print('{:^12}'.format('||') * 3)
    for level in range(height, 0, -1):
        for t in towers:
            try:
                print('{:^12}'.format('==' * t[level-1]), end='')
            except IndexError:
                print('{:^12}'.format('||'), end='')
        print()
    print('-' * 36)
    sleep(1)
````

````python
display()
````

````python
def move(frm, to):
    towers[to].append(towers[frm].pop(-1))
    display()
````

````python
move(0, 2)
````

````python
def hanoi(frm, to, using, levels):
    if levels == 1:
        move(frm, to)
    else:
        hanoi(frm, using, to, levels-1)
        move(frm, to)
        hanoi(using, to, frm, levels-1)
````

#### Beispiel: Mergesort

Ziel: Wir wollen eine Liste von Elementen sortieren, indem wir jedes Mal die Hälfte der Liste sortieren und diese anschließend zusammenführen.

````python
def merge(l1, l2): # O(N), where N is the number of elements in the two lists
    merged = []
    i1 = i2 = 0
    while i1 < len(l1) or i2 < len(l2):
        if i2 == len(l2) or (i1 < len(l1) 
                             and l1[i1] < l2[i2]):
            merged.append(l1[i1])
            i1 += 1
        else:
            merged.append(l2[i2])
            i2 += 1
    return merged
````

````python
l1 = [1, 5, 9]
l2 = [2, 6, 8, 11]
merge(l1, l2)
````

````python
def mergesort(lst):
    if len(lst) <= 1:
        return lst
    else:
        mid = len(lst) // 2
        return merge(mergesort(lst[:mid]), mergesort(lst[mid:]))
````

````python
def insertion_sort(lst):
    for i in range(1, len(lst)):
        for j in range(i, 0, -1):
            if lst[j-1] > lst[j]:
                lst[j-1], lst[j] = lst[j], lst[j-1] # swap
            else:
                break   
````

````python
class Heap:        
    @staticmethod
    def _parent(idx):
        return (idx - 1) // 2
    
    @staticmethod
    def _left(idx):
        return idx*2 + 1

    @staticmethod
    def _right(idx):
        return idx*2 + 2

    def __init__(self, iterable=None):
        if not iterable:
            self.data = []
        else:
            self.data = list(iterable)
            last_internal_idx = Heap._parent(len(self.data)-1)
            for i in range(last_internal_idx, -1, -1):
                self._heapify(i)

    def add(self, x):
        self.data.append(x)
        idx = len(self.data) - 1
        while idx > 0:
            pidx = Heap._parent(idx)
            if self.data[pidx] < self.data[idx]:
                self.data[pidx], self.data[idx] = self.data[idx], self.data[pidx]
                idx = pidx
            else:
                break    
                                
    def max(self):
        assert len(self) > 0
        return self.data[0]

    def _heapify(self, idx):
        while idx < len(self.data):
            lidx = Heap._left(idx)
            ridx = Heap._right(idx)
            maxidx = idx
            if lidx < len(self.data) and self.data[lidx] > self.data[idx]:
                maxidx = lidx
            if ridx < len(self.data) and self.data[ridx] > self.data[maxidx]:
                maxidx = ridx
            if maxidx != idx:
                self.data[idx], self.data[maxidx] = self.data[maxidx], self.data[idx]
                idx = maxidx
            else:
                break

    def pop_max(self):
        assert len(self) > 0
        ret = self.data[0]
        self.data[0] = self.data[-1]
        del self.data[-1]
        self._heapify(0)
        return ret
    
    def __bool__(self):
        return len(self.data) > 0

    def __len__(self):
        return len(self.data)

    def __repr__(self):
        return repr(self.data)

    
def heapsort(iterable):
    h = Heap(iterable)
    ret = []
    while h:
        ret.append(h.pop_max())
    #ret.reverse()
    return ret
````

````python
import timeit
import random

insertionsort_times = []
heapsort_times = []
mergesort_times = []

for size in range(100, 3000, 100):
    insertionsort_times.append(timeit.timeit(stmt='insertion_sort(lst)',
                               setup=f'lst = random.sample(range(1_000_000), {size})',
                               globals=globals(),
                               number=1))
    heapsort_times.append(timeit.timeit(stmt='heapsort(lst)',
                               setup=f'lst = random.sample(range(1_000_000), {size})',
                               globals=globals(),
                               number=1))
    mergesort_times.append(timeit.timeit(stmt='mergesort(lst)',
                               setup=f'lst = random.sample(range(1_000_000), {size})',
                               globals=globals(),
                               number=1))
````
