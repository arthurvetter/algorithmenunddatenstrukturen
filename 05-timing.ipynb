# Zeiten messen

## 0. Vergleich von Algorithmen anhand ihrer Laufzeit

Bevor wir mit der eigentlichen Zeitmessung beginnen, wollen wir anhand eines Beispiels zeigen, dass verschiedene Algorithmen zur Lösung desselben Problems unterschiedlich schnell sein können. Die Effizienz eines Algorithmus lässt sich häufig durch die benötigte Laufzeit vergleichen.

Im folgenden Beispiel berechnen wir die Summe aller Zahlen in einer Liste auf zwei verschiedene Arten: einmal mit einer Schleife und einmal mit der eingebauten Python-Funktion `sum`. Anschließend können wir die Laufzeit beider Ansätze vergleichen.

```python
import random

# Erzeuge eine Liste mit 1 Million Zufallszahlen
numbers = [random.randint(1, 100) for _ in range(100_000_000)]
numbers[:10]
```

```python
# Variante 1: Summe mit einer Schleife berechnen
def sum_with_loop(numbers):
    total = 0
    for number in numbers:
        total += number
    return total

# Variante 2: Summe mit der eingebauten sum()-Funktion berechnen
def sum_with_builtin(numbers):
    return sum(numbers) 
```

```python
print("Summe mit Schleife:", sum_with_loop(numbers))
```

```python
print("Summe mit eingebauter Funktion:", sum_with_builtin(numbers)) 
```

> **Warum ist `sum()` schneller?**  
Die eingebaute Funktion `sum()` ist in C geschrieben und läuft daher viel effizienter als eine Python-Schleife, die vom Interpreter ausgeführt wird.

Aber das müssen wir doch auch genauer hinbekommen. Lassen Sie uns die Zeit mal wirklich messen.

## 1. Zeit messen

Das [`time`](https://docs.python.org/3/library/time.html#module-time) Modul enthält Funktionen, um die aktuelle Systemzeit zu ermitteln und zu interpretieren. Die Systemzeit wird berechnet in vergangenen Sekunden seit dem 01.01.1970, 00:00:00 (UTC)

```python
import time
time.time()
```

Lassen Sie uns die vergangen Jahre berechnen seit dem 01.01.1970:

```python
time.time() // (365 * 24 * 60 * 60)
```

```python
time.localtime(time.time())
```

Indem wir „Zeitstempel“ zu Beginn und am Ende nehmen, können wir die Laufzeit von Code messen:

```python
start = time.time()
time.sleep(1)
end = time.time()
print("Verstrichene Zeit:", end - start)
```

## 2. Vergleich der Laufzeiten zwischen unterschiedlichen Implementierungen

```python
def sum_of_n_2(n):  
    start = time.time()  
    the_sum = 0  
    for i in range(1, n+1):  
        the_sum = the_sum + i  
    end = time.time()  
    return the_sum,end-start
```

Lassen Sie uns jetzt die Funktion einige male ausführen und messen wieviel Zeit diese für die Berechnung der Summe der ersten 10.000 Integers benötigt:

```python
for i in range(5): 
    summe, zeit = sum_of_n_2(10000)
    print(f"Sum is {summe} required {zeit:.7f} seconds")
```

Die Durchführungszeiten sind sehr ähnlich.
Lassen Sie uns jetzt die Funktion einige male ausführen und messen wieviel Zeit diese für die Berechnung der Summe der ersten 1.000.000 Integers benötigt:

```python
for i in range(5): 
    summe, zeit = sum_of_n_2(1000000)
    print(f"Sum is {summe} required {zeit:.7f} seconds")
```

Die Zeiten sind ca. um den Faktor 100 höher. Wir sehen einen linearen Anstieg.
Lassen Sie uns die Summe anders berechnen und messen:

```python
def sum_of_n_3(n):  
    start = time.time()  
    the_sum = (n * (n + 1)) / 2 
    end = time.time() 
    return  the_sum,end-start
```

```python
for i in range(5): 
    summe, zeit = sum_of_n_3(1000000)
    print(f"Sum is {summe} required {zeit:.7f} seconds")
```

Lassen Sie uns diese Funktion nun mit noch größeren Zahlen berechnen. Was fällt Ihnen dabei auf?

```python
for i in range(5): 
    summe, zeit = sum_of_n_3(10000000000000000000)
    print(f"Sum is {summe} required {zeit:.7f} seconds")
```

## 3. Erkenntnisse

- Mit Zeitmessungsbibliotheken können wir das Laufzeitverhalten von Algorithmen systematisch messen.
- Verschiedene Eigenschaften der Eingabe (z. B. zufällig, sortiert, umgekehrt) können einen großen Einfluss auf die Laufzeit von Algorithmen haben.
- Empirische Messungen der Laufzeit liefern nicht immer ein klares, genaues oder konsistentes Bild vom langfristigen Laufzeitverhalten einer Funktion.
- Die Wahl einer falschen Funktionsklasse zur Beschreibung des Laufzeitverhaltens eines Algorithmus kann zu völlig falschen Vorhersagen führen.
- Zeitmessungen sind hilfreich, aber wir brauchen eine systematischere und strengere Methode, um das Laufzeitverhalten von Algorithmen zu beschreiben und zu vergleichen!
